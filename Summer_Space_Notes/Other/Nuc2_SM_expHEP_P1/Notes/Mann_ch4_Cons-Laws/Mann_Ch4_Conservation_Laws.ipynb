## The big idea (chapter thesis)

> 🔑 **Key idea:** *For every symmetry of a system, there is a conserved quantity.* This is **Noether's theorem** (Emmy Noether, 1917).

Chapter 3 set up the *language* of symmetry (groups, algebras, representations). Chapter 4 cashes it in: symmetry isn't just pretty, it **forces conservation laws** — and conservation laws are exactly the things experimentalists can measure. So the logic runs **both ways**:

- **Theory → experiment:** "this theory has symmetry X" ⟹ "quantity Y is conserved, go check it."
- **Experiment → theory:** "we measure Y conserved" ⟹ "the underlying theory should have symmetry X."

> ⚠️ **Subtle point Mann stresses:** In particle physics, a symmetry is a property of the **equations of motion** (the set of *all possible* motions), **not** of any one particular solution. A single solution can look symmetric or not; what matters is whether the *laws* are unchanged.

## 4.1 The Action Principle

The classical-mechanics foundation everything else is built on.

Describe a body by coordinates $q(t) = \{q_1, q_2, \dots, q_n\}$. Define the **action** as the time-integral of the **Lagrangian** $L[q,\dot q]$:

$$S[q] \equiv \int dt\, L[q,\dot q], \qquad \dot q_i = \frac{dq_i}{dt}$$

**The principle:** the true path makes $S$ **stationary** (extremal) under small wiggles $q(t)\to q(t)+\delta q(t)$. Set the first-order variation to zero:

$$0=\delta_{q_i}S=\int_{t_I}^{t_F}dt\left[\frac{\partial L}{\partial q_i}-\frac{d}{dt}\frac{\partial L}{\partial \dot q_i}\right]\delta q_i + \left.\frac{\partial L}{\partial \dot q_i}\delta q_i\right|_{t_I}^{t_F}$$

Two moves make this work:
1. **Integration by parts** turns $\frac{\partial L}{\partial \dot q_i}\delta\dot q_i$ into the $\frac{d}{dt}$ term plus a boundary term.
2. **Fixed endpoints:** $\delta q_i = 0$ at $t_I$ and $t_F$ (we compare paths with the *same* start and end), so the boundary term dies.

Since $\delta q_i$ is otherwise arbitrary, the bracket must vanish:

$$\boxed{\;\frac{\partial L}{\partial q_i}-\frac{d}{dt}\frac{\partial L}{\partial \dot q_i}=0\;}\qquad\text{Euler–Lagrange equations}$$

Defining the **canonical momentum** $p_i \equiv \dfrac{\partial L}{\partial \dot q_i}$, this is just:

$$\frac{dp_i}{dt}=\frac{\partial L}{\partial q_i}$$

> 🔑 **Why the action is the right tool:** It's a single **scalar**. Scalars are ideal for talking about symmetry because "the action is unchanged" is a clean, observer-independent statement. The same principle generalizes to QM, GR, and QFT — essentially all of fundamental physics is built on an action principle.

**Invariance ⟹ covariance.** If a group of transformations $q_i' = g_i[q]$ leaves the action invariant, $S[q']=S[q]$, then whenever $q(t)$ solves Euler–Lagrange, the transformed $q'(t')$ *also* solves a set of equations of motion — obtained from the originals by the same transformation. (The group structure matters: every element needs an inverse so the Jacobian $\det\frac{\delta q_j'}{\delta q_i}\neq 0$.) In short: **an invariant action yields covariant equations of motion.**

### §4.1 — expanded explanation of the book snippet (pp. 65–66)

Snippet sent during the session:

![Mann pp. 65-66 snippet](fig_4.1_page65-66_snippet.png)

**Plain-language breakdown of this passage:**

**1. Canonical momentum & what Euler–Lagrange means.**
$$p_i \equiv \frac{\partial L}{\partial \dot q_i}$$
This is just a *definition*: the "momentum" attached to coordinate $q_i$ is how much $L$ changes when you change that coordinate's velocity. For a normal particle $L=\tfrac12 m\dot q^2$, giving $p=m\dot q$ — ordinary momentum. "Canonical" just means "the general version."

The takeaway of §4.1 is an **if and only if**: solving Euler–Lagrange (4.3) $\Leftrightarrow$ finding the path that extremizes the action. A path that solves (4.3) extremizes $S$; a path that doesn't, doesn't.

**2. Why the action is a big deal.** $S$ is a single **number** (a scalar) computed from a whole path. "Make $S$ extremal" produces *all* the equations of motion, and the same trick runs classical mechanics, QM, GR, and QFT. Being a scalar makes it perfect for asking "is this symmetric?" — you only check whether one number stays the same.

**3. The symmetry argument (eqs. 4.5–4.6) — the heart of the passage.**
A symmetry is a group $\mathcal G=\{g_i\}$ that leaves the action unchanged: $q_i' = g_i[q]$ with $S[q'] = S[q]$ (*invariance*). Question: if we rewrite the motion in the new coordinates $q'$, do they obey equations of motion? Using the chain rule (factor $\delta q_j'/\delta q_i$) and the invariance $S[q']=S[q]$, the variation collapses to
$$\delta_{q_i'}S[q']=0,$$
so the new coordinates **also** extremize the action and obey equations of motion.

> 🔑 **One-line conclusion (top of p. 66):** an **invariant action** ⟹ **covariant equations of motion.** The new equations aren't identical to the old, but have the *same form* — you get them by applying the same transformation. That's what "covariant" means: they transform *with* you.

**4. The fine print — why $\mathcal G$ must be a *group*.** The argument needs the chain-rule factor nonzero: $\det\frac{\delta q_j'}{\delta q_i}\neq 0$, i.e. the transformation is **invertible** — exactly the group axiom "every element has an inverse."

> ⚠️ **Why it matters:** if that determinant were zero (no inverse), you couldn't conclude $\delta_{q'}S[q']=0$; you'd only get that *some linear combination* of variations vanishes — not enough to pin down equations of motion for $q'$. Group invertibility is what makes the symmetry → covariance logic airtight.

**In one breath:** *Canonical momentum is just $\partial L/\partial\dot q$. Solving Euler–Lagrange = extremizing the action. If a symmetry leaves the action's value unchanged, the transformed coordinates automatically satisfy equations of motion of the same form — and this works only because symmetry transformations are invertible (a group), so the chain-rule factor never vanishes.*

## 4.2 Noether's Theorem (the engine)

> 📄 A fuller, typeset version of §4.2–4.3 lives in the companion file **`Mann_4.2_4.3_Noether_and_Currents.pdf`** (built this session).

The trick: look at a **fully arbitrary** variation — one where neither the equations of motion hold *nor* the endpoints are fixed. Then

$$\delta_{q_i}L=\left[\frac{\partial L}{\partial q_i}-\frac{d}{dt}\frac{\partial L}{\partial\dot q_i}\right]\delta q_i+\frac{d}{dt}\big[p_i\,\delta q_i\big]$$

Specialize $\delta q_i$ to a **small symmetry transformation** with parameters $\alpha=\{\alpha_1,\dots,\alpha_m\}$, where $\alpha=0$ is the identity:

$$\delta_{\alpha_j}q_i=\left.\frac{\partial q_i'}{\partial\alpha_j}\right|_{\alpha=0}\delta\alpha_j$$

> 🔑 **The pivotal fact:** For a *symmetry*, $\delta_\alpha L$ must be a **total time derivative** (so it integrates to zero and leaves $S$ invariant). For a generic variation it is *not* a total derivative unless the equations of motion hold. This is the whole difference.

Write that total derivative as $\delta_{\alpha_j}L=\frac{d}{dt}G$, where $G=G(q,\partial q/\partial\alpha)$ vanishes at the endpoints and is worked out **case by case**. Then build

$$\boxed{\;J_k=p_i\frac{\partial q_i'}{\partial\alpha_k}-G\;}\qquad\text{Noether current}$$

A few lines (using Euler–Lagrange + the total-derivative property) give $\frac{d}{dt}J_k=0$.

> 🔑 **Noether's theorem:** Each symmetry parameter $\alpha_k$ gives a conserved current $J_k$. What $J_k$ *means* depends entirely on what the symmetry is.

> ⚠️ **Two easy-to-miss points:** (1) Two *different* variations are used — the master identity comes from a sloppy variation (EOM not assumed); the final $dJ_k/dt=0$ is evaluated *on* a real solution (EOM assumed). (2) $G$ is not optional: it's $0$ for translations/rotations but $G=L$ for time translation.

## 4.3 Spacetime Symmetries and Their Noether Currents

Three cases, each with $G$ chosen appropriately. Turn the crank of $J_k=p_i\,\partial q_i'/\partial\alpha_k-G$ three times.

| Symmetry | Transformation | $G$ | Noether current $J_k$ | Conserved quantity |
|---|---|---|---|---|
| **4.3.1 Spatial translation** | $q\to q+\alpha$, so $L[q+\alpha,\dot q]=L[q,\dot q]$ | $0$ | $J_k=p_i\delta_{ik}=p_k$ | **Linear momentum** $\vec p$ |
| **4.3.2 Rotation** $SO(3)$ | $q_i'=q_i-(\vec\alpha\times\vec q)_i$ | $0$ | $J_k=p_i\,\epsilon_{ikj}q_j=(\vec q\times\vec p)_k$ | **Angular momentum** $\vec q\times\vec p$ |
| **4.3.3 Time translation** | $t'=t+\delta\alpha$, $\partial q_i'/\partial\alpha=\dot q_i$ | $G=L$ | $J=p_i\dot q_i-L\equiv H$ | **Energy** (Hamiltonian $H$) |

The three pillars:

> - **Space-translation invariance ⟹ conservation of linear momentum**
> - **Rotational invariance ⟹ conservation of angular momentum**
> - **Time-translation invariance ⟹ conservation of energy**

> ⚠️ **Details:** For translations/rotations $\partial L/\partial\alpha|_0=0$, so $G=0$ (a constant $G$ would just shift the conserved quantity by an irrelevant constant). For rotations $L$ depends only on $|\vec q|$ (magnitudes are rotation-invariant). **Energy is conserved for any *closed* system** — one whose Lagrangian has no *explicit* time dependence; explicit $t$-dependence would let something external modify the system later, breaking time-translation invariance.

## 4.4 Symmetries and Quantum Mechanics

The transition to QM changes things in **two ways**:
1. The symmetry↔conservation relationship gets **reformulated** (operators & commutators instead of currents).
2. It opens the door to **internal symmetries** — symmetries that rearrange charges / internal wavefunction structure *without* touching space and time (e.g. shuffling quark colors). These underlie all the non-gravitational forces.

**Setup.** A state is a wavefunction $\Psi$ obeying $i\hbar\,\partial_t\Psi=H\Psi$, where now the **Hamiltonian** $H$ (built from the classical $H=p_i\dot q_i-L$) plays the fundamental role — not the Lagrangian.

A symmetry acts as a matrix on the wavefunction:
$$q_i'=g_i[\alpha,q]\;\longrightarrow\;\Psi'(t;q')=U(\alpha)\,\Psi(t;q)$$

**Symmetries must be unitary.** Normalization can't change under a symmetry, so
$$\int dq'\,\Psi'^\dagger\Psi'=\int dq\,\Psi^\dagger U^\dagger U\Psi=1\;\Longrightarrow\;\boxed{U^\dagger U=I}$$

**Quantum Noether.** For $\omega=\int dq\,\Psi^\dagger U\Psi$, differentiating in time and using Schrödinger's equation:
$$\frac{d\omega}{dt}=\frac{i}{\hbar}\int dq\,\Psi^\dagger[H,U]\Psi$$
$$\boxed{\;\text{conserved quantum number }\omega\;\Longleftrightarrow\;[U,H]=0\;\Longleftrightarrow\;U\text{ is a symmetry}\;}$$

The set $\{U(\alpha_i)\}$ forming a group is the **symmetry group of the Hamiltonian**. Crucially, $U$ need **not** be spacetime-related — it can just mix wavefunction components (internal symmetry).

**The three spacetime cases in QM:**

| § | Build $U$ from… | Operator | Commutator condition |
|---|---|---|---|
| **4.4.1 Spatial translation** | $\Psi'(q)=\Psi(q-\alpha)\approx(1-\vec\alpha\cdot\vec\nabla)\Psi$ | $\vec P=-i\hbar\vec\nabla$ | $[\vec P,H]=0$, and $[P_i,P_j]=0$ |
| **4.4.2 Rotation** | $\Psi'(q)=\Psi(R^{-1}q)\approx(1+i\vec\alpha\cdot(\vec q\times(-i\vec\nabla)))\Psi$ | $\vec L=\vec q\times(-i\hbar\vec\nabla)$ | $[\vec L,H]=0$, but $[L^a,L^b]=i\hbar\epsilon^{abc}L^c\neq0$ |
| **4.4.3 Time translation** | $\Psi'(t)=\Psi(t-\alpha)\approx(1+\frac{i}{\hbar}\alpha H)\Psi$ | $H$ | $[H,H]=0$ trivially ⟹ energy **always** conserved in QM |

> ⚠️ **The key quantum twist (rotations):** Momentum components all commute ($\nabla_i\nabla_j=\nabla_j\nabla_i$), so in a translation-invariant system you can define $P_x,P_y,P_z$ **simultaneously**. But the angular-momentum components **don't** commute:
> $$[L^a,L^b]=i\hbar\,\epsilon^{abc}L^c$$
> (Mann grinds out $[L_x,L_y]=i\hbar L_z$ explicitly.) So even though *classically* $\vec q\times\vec p$ is conserved about every axis, **quantum-mechanically angular momentum can only be defined about one chosen axis** — once you pick it, the other components become unmeasurable. This is the same algebra the rotation generators $J$ satisfied in Ch. 3, eq. (3.13).

## 4.5 Summary

Under a symmetry $g[\alpha]$:

| Framework | Condition | Conserved object |
|---|---|---|
| **Classical** | invariance of the **action** | $J_k=p_i\dfrac{\partial q_i'}{\partial\alpha_k}-G\Big|_{\alpha_k=0}$ |
| **Quantum** | **commutation** $[U(\alpha),H]=0$ | $\omega=\displaystyle\int dq\,\Psi^\dagger U\Psi$ |

For spacetime symmetries specifically:
- Rotational invariance ⟹ conservation of **angular momentum**
- Space-translation invariance ⟹ conservation of **linear momentum**
- Time-translation invariance ⟹ conservation of **energy**
