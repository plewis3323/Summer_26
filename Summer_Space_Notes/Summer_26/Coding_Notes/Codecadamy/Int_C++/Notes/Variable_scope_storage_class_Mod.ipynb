- Intro to Variable Scope: 

The term variable scope refers to the region or section of code where a variable can be accessed or is visible. This means that the part of the code where a variable can be used is determined by its scope.

In C++, there are different types of scopes, each determining how and where a variable can be accessed:

Global scope refers to variables declared outside all functions
or blocks, making them accessible throughout the entire program.
Local scope, on the other hand, applies to variables declared within a specific function, limiting their visibility to that function.
Block scope is even more restrictive, covering variables declared within a set of braces {}, such as those inside 
loops or conditional statements.

We will learn about these scope types in the upcoming exercises.

Understanding variable scope allows us to control the visibility and lifetime of variables in our program, preventing potential 
errors and enhancing code readability.


- Understanding Local and Global Scope: 


Global variables: Declared outside all functions, accessible everywhere.
Local variables: Declared inside a function or block, accessible only there.
Local variables with the same name as global variables will override the global variable inside their scope.
Accessing a local variable outside its function causes an error.
Overusing global variables can make code harder to manage and debug.



ex.)

  cout << "Player score from inside the showStats() function: " << playerScore << endl;
}


int main() {
  //int playerScore = 50; 
  cout << "Player score from inside the main() function: " << playerScore << endl;
  showStats();

- Block Scope: 


Block scope means a variable is only accessible within the curly braces {} where it is declared. If a variable is declared inside a block, 
it cannot be used outside that block. Variables outside the block can still be used inside the block. If a variable with the same name exists 
inside and outside the block, the inner one is used inside the block.

- Introduction to Storage Classes: 


Automatic (auto) variables are declared inside a function and exist only during its execution.
Static (static) variables are initialized only once and retain their value even after the function exits.
Extern (extern) variables are accessible across different files in a program.

# Static Variables in Functions:

- **`static` keyword** — declares a variable that *keeps its value between function calls* instead of being recreated each time.
  - A **static variable** preserves its value even after the function where it was declared finishes executing.
    - **Regular variable:** created and destroyed each time the function is called.
    - **Static variable:** maintains its value throughout the program's execution.
  - **Use cases** — remembering something across multiple calls: e.g. counting how many times a function ran, or caching a result.

  **Declaration syntax:**
  ```cpp
  static dataType variableName;

  Example — count() with a static variable num:
  void count() {
      static int num = 0;   // static variable
      num++;
      cout << num << endl;
  }

  - The = 0 initialization runs only on the first call; afterward it is skipped.
  - num does not reset — it retains and increments its value across calls:
  count();  // Outputs 1
  count();  // Outputs 2
  count();  // Outputs 3

  Key points:
  - Static variables have unique storage.
  - Their lifetime extends throughout the program.

````python
%%cpp
// Ex.)

#include <iostream>
using namespace std;

void playerHealth() {
  static int health = 100;
  health -= 10;
  cout << "Current health: " << health << endl;
}

int main() {
  playerHealth();
  playerHealth();
  return 0;
}
````

# Static Members in Class:



  **First — what is an instance?**

  - A class is a blueprint (e.g. the design for a house).
  - An instance (also called an object) is one actual thing built from that blueprint (one specific house).
  - From one class you can make many instances, and each instance normally has its own separate copy of the class's variables.

  UserSession userA;   // instance 1
  UserSession userB;   // instance 2  <- a different object

  The idea of static members

  - A static member belongs to the class itself, not to any one instance.
  - Instead of every instance having its own copy, all instances share one single copy.
  - Two kinds:
    - Static variable -> one shared piece of data for the whole class (e.g. a running total).
    - Static function -> a function you call on the class without needing an instance (e.g. a utility/math helper).

  Declaring static members:

    class SomeClass {
    public:
      static int someVariable;
      static int someFunction(int param1);
    };
  Accessing them — use the class name and :: (scope operator), not an object:

    SomeClass::someVariable;
    SomeClass::someFunction(2);
  Worked example — counting logged-in users

  - Goal: track how many users are logged in across the whole app.
    Each user = one UserSession instance. The shared count = a static int loggedInUsers:
    
        class UserSession {
        public:
        static int loggedInUsers;
        UserSession();                     // constructor (runs when an instance is created)
        ~UserSession();                    // destructor  (runs when an instance is destroyed)
        static void displayLoggedInUsers();
        };

  Implementation:

    int UserSession::loggedInUsers = 0;          // initialize the shared counter

    UserSession::UserSession() { loggedInUsers++; }   // a user logs in  -> +1
    UserSession::~UserSession() { loggedInUsers--; }   // a user logs out -> -1

    void UserSession::displayLoggedInUsers() {
      cout << "Total logged-in users: " << loggedInUsers << endl;
    }
  - Create an instance -> constructor runs -> count goes up.
  - Destroy an instance -> destructor runs -> count goes down.
  - A static function is defined just like a normal function.

  Important rule — initialize static variables outside the class

   A static variable must be given its starting value outside the class definition:

     int UserSession::loggedInUsers = 0;   // correct - done outside the class
  Doing it inside the class causes a compile error:

    class UserSession {
    public:
      static int loggedInUsers = 0;   // NOT allowed
      };

  Calling the static function:

    UserSession::displayLoggedInUsers();
  Don't access static members through an instance

Because static members belong to the class, reaching them via an object (with the dot .) is bad practice:

    int main() {
      UserSession someUser;
      cout << someUser.loggedInUsers << endl;   // compiles, but misleading
     }
  
  It works, but it looks like per-object data when it's really shared — so always use ClassName::member instead.

  Key takeaways

  - Static member = shared by the whole class, not per-instance.
  - Access with ClassName::member.
  - Static variables must be initialized outside the class.
  - Great for: a count shared across all objects, or utility functions (e.g. a Math class).

# External Linkage with extern:

 The problem it solves

  - Big programs are split across multiple source files to stay readable.
  - But what if a variable or function defined in one file needs to be used in another?
  - The extern keyword is the answer. It tells the compiler: "This variable/function is defined somewhere else — but I'm going to use it here."

  Basic syntax

  Declaring a variable as external:

  extern type variableName;
  - Note: extern only works with global variables or functions (not local ones).
  - Common alternative (and usually preferred): put declarations in a header file and #include it wherever needed.

  Example — a shared logging system

  Goal: a global logLevel controls which messages show, and logMessage() prints them. Defined in one file, used in many.

  In Logger.cpp — define the variable and function:

  int logLevel = 2; // 1: Error, 2: Warning, 3: Info

  void logMessage(int level, const string &message) {
      if (level <= logLevel) {
          cout << "Log (Level " << level << "): " << message << endl;
      }
  }
  In main.cpp — declare them with extern so this file knows they exist elsewhere:

  extern int logLevel;
  extern void logMessage(int level, string &message);
  - extern here says: "logLevel and logMessage() live in another file (Logger.cpp), but I'll use them in main.cpp."
  - Define vs declare: the actual = 2 and function body live in one file (the definition); other files just extern-declare them (a promise they exist).

  Drawback to remember

  - An extern variable is a shared global — every file points to the same memory location.
  - So changing it in one place affects every other part of the program that uses it. Use with care.

  Using extern with precompiled code

  - extern can also reference functions/classes from already-compiled object files (code someone else wrote and compiled — you don't have the source).

  -   extern class ExternalClass {
    public:
    void performTask();
    };
  - For this to work, the object file or library must be linked during compilation — you point the linker at it.

  Example linker command with g++:

  g++ main.cpp -o main -L/path/to/library -lmyLibrary
  - -L/path/to/library → the directory where the library lives.
  - -lmyLibrary → links the library itself (a file named like libmyLibrary.a).
  - myLibrary and /path/to/library are placeholders — swap in your real names. (e.g. library libexample.a in /usr/local/lib → -L/usr/local/lib -lexample.)

  Once declared and linked, you use it like normal code:

  ExternalClass obj;
  obj.performTask();
  Key takeaways

  - extern = "defined elsewhere, used here" — links across files.
  - Works only with globals and functions.
  - One file defines (with a value/body); other files declare with extern.
  - Watch out: extern globals are shared memory — a change affects everyone.
  - Also links precompiled libraries via the linker flags -L (path) and -l (library).

# Read-Only Functions:


 What they are

  - A read-only function is a class function that promises not to modify any class member.
  - You mark it by putting const after the parameter list:

  int someFunction() const {
      // function body
  }
  - The const keyword prevents accidental changes to the object's data — making code more predictable and reliable.

  What const actually blocks

  Inside a const function you can read members but not change them:

  class SomeClass {
  private:
      int someVariable = 100;

  public:
      void someFunction() const {
          someVariable++;                 // ERROR! can't modify a member
          cout << someVariable << endl;   // OK - just reading is fine
      }
  };
  - Modify a member → compile error.
  - Read/print a member → perfectly allowed.

  Why they're useful

  - They let you expose an object's data without letting anyone alter it — a safety guarantee.
  - Example: a Book class with a name. A printName() function should only display the name, never change it. Marking it const guarantees name stays intact.

  -   class Book {
    private:
    string name;
    public:
    Book(string bookName) : name(bookName) {}   // constructor (initializer list)
    void printName() const {
        cout << name << endl;
    }
    };
  - printName() const → "this function only reads name, it will never modify the object."

  Key takeaways

  - Put const after the parentheses to make a function read-only: void f() const { ... }.
  - Inside it: reading members = OK, changing members = compile error.
  - Use it for any function that should only look at data (getters, printers, etc.) — it documents intent and enforces it.

# The mutable Keyword:

 The problem it solves

  - A const (read-only) function can't modify any member of the class.
  - But sometimes you need a read-only function to change one specific member anyway.
  - mutable is the exception: it marks a member that can still be changed even inside a const function.

  Syntax:

  mutable DataType variableName;
  Example — counting calls to printName()

  Recall the Book class. Say we want to track how many times printName() has been called using a counter member.

  - Problem: printName() is const, so counter++ would normally cause a compile error.
  - Solution: declare counter as mutable — now it's allowed to change inside the const function.

  -   mutable int counter;

  Updated Book class:

  class Book {
  private:
      string name;
      mutable int counter;

  public:
      Book(string bookName) : name(bookName), counter(0) {}
      void printName() const {
          counter++;                 // allowed, because counter is mutable
          cout << name << endl;
      }
  };
  How it fits together

  - printName() const → promises not to modify the Book object...
  - ...except for members marked mutable.
  - So counter++ works, while name is still protected — you get the read-only guarantee and your counter.

  Key takeaways

  - mutable = "this member can change even inside a const function."
  - Declare it on the member: mutable int counter;.
  - Use it for "bookkeeping" data (call counts, caches, timestamps) that doesn't really count as the object's core state.
  - Everything not marked mutable stays protected by const.
