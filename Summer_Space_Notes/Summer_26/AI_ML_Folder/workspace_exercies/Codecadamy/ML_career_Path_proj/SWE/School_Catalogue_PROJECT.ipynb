# School Catalogue

*Codecademy MLE Path — Python classes / inheritance.*

The NYC Department of Education wants a digital catalogue holding quick reference material for
every school in the city. Primary, middle, and high schools share most of their data, so they all
inherit from one parent `School` class and each child adds only what makes it different.

| class | inherits | adds |
|---|---|---|
| `School` | — | `name`, `level`, `numberOfStudents` + getters, one setter, `__repr__` |
| `PrimarySchool` | `School` | `pickupPolicy` |
| `MiddleSchool` | `School` | nothing |
| `HighSchool` | `School` | `sportsTeams` |

**This notebook holds your submitted code**, run top to bottom, with a check after each part and a
review section at the end.


---
## Part 1 — The `School` parent class

**Steps 1-3.** Class named `School`, constructor taking `self, name, level, numberOfStudents`, each
stored on `self`.

**Steps 4-5.** Getters for all three properties; a setter for `numberOfStudents` only.

**Step 6.** `__repr__` returning `A {level} school named {name} with {numberOfStudents} students`.

The setter is the reason `numberOfStudents` is different from the other two: a school's name and
level don't change, but enrollment does every year.


````python
class School: 
  def __init__(self, name, level, numberOfStudents):
    self.name = name 
    self.level =  level 
    self.numberOfStudents = numberOfStudents 
  def getname(self):
    return self.name 
  def getlevel(self):
    return self.level 
  def getnumberOfStudents(self):
    return self.numberOfStudents
  def set_number_of_students(self, numberOfStudents):
    self.numberOfStudents = numberOfStudents
  def __repr__(self):
    return f"A {self.level} school named {self.name} with {self.numberOfStudents} students."
````

**Step 7.** Test it — build a `School`, read it with the getters, change enrollment with the setter, print it.

````python
a = School("Codecademy", "high", 100)
print(a)
print(a.getname())
print(a.getlevel())
a.set_number_of_students(200)
print(a.getnumberOfStudents())
# expected:
# A high school named Codecademy with 100 students.
# Codecademy
# high
# 200
````

### Check — Part 1

````python
def check_part1():
    s = School("Test School", "middle", 50)
    assert s.name == "Test School" and s.level == "middle" and s.numberOfStudents == 50
    assert s.getname() == "Test School", "getname() should return the name"
    assert s.getlevel() == "middle", "getlevel() should return the level"
    assert s.getnumberOfStudents() == 50
    s.set_number_of_students(75)
    assert s.getnumberOfStudents() == 75, "setter should overwrite numberOfStudents"
    assert isinstance(repr(s), str), "__repr__ must return a string"
    assert "middle" in repr(s) and "Test School" in repr(s) and "75" in repr(s)
    print("Part 1 OK")
````

## Part 2 — `PrimarySchool`

**Steps 8-11.** Inherits from `School`. Constructor takes `self, name, numberOfStudents,
pickupPolicy` — no `level` parameter, because a primary school's level is never anything but
`'primary'`. That constant gets hardcoded into the `super().__init__()` call.

Argument order matters: the parent signature is `(name, level, numberOfStudents)`, so the string
`"primary"` goes in the **middle** slot.

**Steps 12-13.** One new getter, and a `__repr__` that reuses the parent's via `super().__repr__()`
rather than retyping it. `super().__repr__()` *returns* a string, so you concatenate onto it.


````python
class PrimarySchool(School):
  def __init__(self, name, numberOfStudents, pickupPolicy):
    super().__init__(name, "primary", numberOfStudents)
    self.pickupPolicy = pickupPolicy 
  def getpickupPolicy(self):
    return self.pickupPolicy
  def __repr__(self):
    return super().__repr__() + f". The pickup policy is {self.pickupPolicy}"
````

**Step 14.** Test the new constructor, getter, and `__repr__`.

````python
b = PrimarySchool("Codecademy", 300, "Pickup Allowed")
print(b.getpickupPolicy())
print(b)
# expected:
# Pickup Allowed
# A primary school named Codecademy with 300 students.. The pickup policy is Pickup Allowed
````

### Check — Part 2

````python
def check_part2():
    p = PrimarySchool("Test Primary", 300, "Pickup after 3pm")
    assert isinstance(p, School), "PrimarySchool must inherit from School"
    assert p.level == "primary", "level should be hardcoded to 'primary'"
    assert p.name == "Test Primary" and p.numberOfStudents == 300, "args passed to super() in the wrong order?"
    assert p.getpickupPolicy() == "Pickup after 3pm"
    assert p.getname() == "Test Primary", "inherited getters should still work"
    assert "Pickup after 3pm" in repr(p), "__repr__ should mention the pickup policy"
    assert "primary school named Test Primary" in repr(p), "__repr__ should build on super().__repr__()"
    print("Part 2 OK")
````

## Part 3 — `HighSchool`

**Steps 15-16.** Same shape as `PrimarySchool` — only the hardcoded level, the extra property, and
the `__repr__` tail differ. That symmetry is the point of the exercise: once the parent holds the
shared work, each child is about four lines.

`sportsTeams` is a **list**, so printing it inside an f-string shows the brackets and quotes.


````python
class HighSchool(School):
  def __init__(self, name, numberOfStudents, sportsTeams):
    super().__init__(name, "high", numberOfStudents)
    self.sportsTeams = sportsTeams

  def get_sports_teams(self):
    return self.sportsTeams

  def __repr__(self):
    return super().__repr__() + f". The sports teams are {self.sportsTeams}"
````

**Step 16.** Test the high school.

````python
c = HighSchool("Codecademy High", 500, ["Tennis", "Basketball"])
print(c.get_sports_teams())
print(c)
# expected:
# ['Tennis', 'Basketball']
# A high school named Codecademy High with 500 students.. The sports teams are ['Tennis', 'Basketball']
````

### Check — Part 3

````python
def check_part3():
    h = HighSchool("Test High", 500, ["basketball", "tennis"])
    assert isinstance(h, School), "HighSchool must inherit from School"
    assert h.level == "high", "level should be hardcoded to 'high'"
    assert h.name == "Test High" and h.numberOfStudents == 500
    assert h.get_sports_teams() == ["basketball", "tennis"]
    assert "basketball" in repr(h), "__repr__ should mention the sports teams"
    assert "high school named Test High" in repr(h), "__repr__ should build on super().__repr__()"
    print("Part 3 OK")
````

## Review — two things to clean up

The submitted code passes every check. Two issues are worth fixing anyway, both visible in the
printed output above.

### 1. Double period in the child `__repr__`s

`School.__repr__` ends its string with `students.` and each child then appends `". The pickup
policy is ..."`. Concatenating them gives:

```
A primary school named Codecademy with 300 students.. The pickup policy is Pickup Allowed
                                                   ^^
```

Two sentences, three periods. The fix is to pick one owner of the punctuation — drop the trailing
period from the parent and let each child start its addition with `". "`. The parent still reads
fine on its own, and every child now joins cleanly.

**The general lesson:** when a method is designed to be extended by subclasses, its return value has
to compose. Trailing punctuation is a decision about the *end of the sentence*, and a parent class
doesn't know whether it's the end.

### 2. `MiddleSchool` is missing

The narrative lists three child classes — Primary, Middle, and High — but the numbered tasks only
walk through two, so it's easy to stop at `HighSchool`. `MiddleSchool` adds no properties, no
getters, and no `__repr__` override; the entire class is a constructor that forwards to `super()`
with `"middle"` hardcoded. That is what makes it a good demonstration: inheritance means an
"empty" subclass still gets all six of the parent's members for free.

### Also worth noting: getter naming

`getname` / `getlevel` / `getnumberOfStudents` sit next to `set_number_of_students` and
`get_sports_teams` — three different conventions in one file. Nothing breaks, but Python's is
snake_case (`get_name`, `get_level`, `get_number_of_students`). Consistency here is what stops you
guessing at a method name six months from now.


Corrected version, with all three fixes. Everything else is unchanged from the submission.

````python
class School:
  def __init__(self, name, level, numberOfStudents):
    self.name = name
    self.level = level
    self.numberOfStudents = numberOfStudents

  def get_name(self):
    return self.name

  def get_level(self):
    return self.level

  def get_number_of_students(self):
    return self.numberOfStudents

  def set_number_of_students(self, numberOfStudents):
    self.numberOfStudents = numberOfStudents

  def __repr__(self):
    # no trailing period - each subclass continues the sentence
    return f"A {self.level} school named {self.name} with {self.numberOfStudents} students"


class PrimarySchool(School):
  def __init__(self, name, numberOfStudents, pickupPolicy):
    super().__init__(name, "primary", numberOfStudents)
    self.pickupPolicy = pickupPolicy

  def get_pickup_policy(self):
    return self.pickupPolicy

  def __repr__(self):
    return super().__repr__() + f". The pickup policy is {self.pickupPolicy}"


class MiddleSchool(School):
  # no extra properties, getters, or methods - everything comes from School
  def __init__(self, name, numberOfStudents):
    super().__init__(name, "middle", numberOfStudents)


class HighSchool(School):
  def __init__(self, name, numberOfStudents, sportsTeams):
    super().__init__(name, "high", numberOfStudents)
    self.sportsTeams = sportsTeams

  def get_sports_teams(self):
    return self.sportsTeams

  def __repr__(self):
    return super().__repr__() + f". The sports teams are {self.sportsTeams}"
````

The catalogue the Department of Education actually asked for — one list, three kinds of school, each printing itself correctly.

````python
catalogue = [
  PrimarySchool("Codecademy Primary", 300, "Pickup after 3pm"),
  MiddleSchool("Codecademy Middle", 400),
  HighSchool("Codecademy High", 500, ["Tennis", "Basketball"]),
]

for school in catalogue:
  print(school)
````

### Check — corrected version

````python
def check_fixed():
    p = PrimarySchool("P", 300, "Pickup after 3pm")
    m = MiddleSchool("M", 400)
    h = HighSchool("H", 500, ["tennis"])
    assert ".." not in repr(p) and ".." not in repr(h), "double period still present"
    assert repr(p) == "A primary school named P with 300 students. The pickup policy is Pickup after 3pm", f"got: {p!r}"
    assert repr(m) == "A middle school named M with 400 students", f"got: {m!r}"
    assert repr(h) == "A high school named H with 500 students. The sports teams are ['tennis']", f"got: {h!r}"
    assert m.get_name() == "M" and m.get_level() == "middle", "MiddleSchool inherits every getter"
    assert p.get_pickup_policy() == "Pickup after 3pm"
    assert h.get_sports_teams() == ["tennis"]
    print("Corrected version OK")

check_fixed()
````

## TL;DR

- A **parent class holds what's shared**; each child adds only its difference. `School` carries
  three properties, three getters, a setter, and a `__repr__`; every child is then ~4 lines.
- **A child with one fixed value for a parent property doesn't take it as a parameter** — it
  hardcodes it in the `super().__init__()` call. That's why `PrimarySchool(name, numberOfStudents,
  pickupPolicy)` has no `level` argument.
- **`super().__init__()` is positional** — arguments must line up with the parent's signature:
  `super().__init__(name, "primary", numberOfStudents)`, not the other order.
- **`super().__repr__()` returns a string, it doesn't print.** Concatenate onto it to extend the
  parent's output instead of retyping it.
- **A method meant to be extended shouldn't own the final punctuation** — that caused the `..` here.
- **A subclass that adds nothing is still worth writing.** `MiddleSchool` gets all six inherited
  members from a two-line constructor.
- Getter methods must not share a name with the attribute they wrap — `self.name` and `def name()`
  collide. Hence the `get_` prefix.
