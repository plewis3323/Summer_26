## Part 1 — Making the Menus

**Steps 1–4.** Create a `Menu` class with a constructor taking `self, name, items, start_time, end_time`
and storing each as an instance attribute.

- `name` — string, e.g. `'brunch'`
- `items` — dict of `item name -> price`
- `start_time` / `end_time` — ints, the hours the menu is served

**Step 8.** Add a `__repr__` that returns the menu name and the hours it's available, e.g.
`brunch menu available from 11 am to 4 pm`.

**Step 10.** Add `calculate_bill(self, purchased_items)` that takes a list of item names and returns
the total price of those items.

Come back and grow this one cell as you go — re-run it after each addition.

```python
class Menu:
  def __init__(self, name, items, start_time, end_time):
    self.name = name
    self.items = items
    self.start_time = start_time
    self.end_time = end_time

  # helper: 24-hour int -> "11 am" / "4 pm"
  def format_time(self, hour):
    if hour == 12:
      return "12 pm"
    elif hour > 12:
      return str(hour - 12) + " pm"
    else:
      return str(hour) + " am"

  def __repr__(self):
    return self.name + " menu available from " + self.format_time(self.start_time) + " to " + self.format_time(self.end_time)

  def calculate_bill(self, purchased_items):
    total = 0
    for item in purchased_items:
      total += self.items[item]
    return total
```

**Steps 5–7.** Create the four menus. The item dicts are given (Codecademy supplies these) — your job
is to wrap each one in a `Menu` object with the right hours.

| variable | `name` string | hours |
|---|---|---|
| `brunch` | `'brunch'` | 11 am – 4 pm |
| `early_bird` | `'early-bird'` | 3 pm – 6 pm |
| `dinner` | `'dinner'` | 5 pm – 11 pm |
| `kids` | `'kids'` | 11 am – 9 pm |

```python
brunch_items = {'pancakes': 7.50, 'waffles': 9.00, 'burger': 11.00, 'home fries': 4.50, 'coffee': 1.50, 'espresso': 3.00, 'tea': 1.00, 'mimosa': 10.50, 'orange juice': 3.50}

brunch = Menu("brunch", brunch_items, 11, 16)

early_bird_items = {
  'salumeria plate': 8.00, 'salad and breadsticks (serves 2, no refills)': 14.00, 'pizza with quattro formaggi': 9.00, 'duck ragu': 17.50, 'mushroom ravioli (vegan)': 13.50, 'coffee': 1.50, 'espresso': 3.00,
}
early_bird = Menu("early-bird", early_bird_items, 15, 18)

dinner_items = {'crostini with eggplant caponata': 13.00, 'caesar salad': 16.00, 'pizza with quattro formaggi': 11.00, 'duck ragu': 19.50, 'mushroom ravioli (vegan)': 13.50, 'coffee': 2.00, 'espresso': 3.00,
}

dinner = Menu("dinner", dinner_items, 17, 23)

kids_items = {'chicken nuggets': 6.50, 'fusilli with wild mushrooms': 12.00, 'apple juice': 3.00}

kids = Menu("kids", kids_items, 11, 21)
```

**Step 9 & 11–12.** Check your `__repr__` and `calculate_bill`.

```python
print(brunch)
# expected: brunch menu available from 11 am to 4 pm

print(brunch.calculate_bill(['pancakes', 'home fries', 'coffee']))
# expected: 13.5

print(early_bird.calculate_bill(['salumeria plate', 'mushroom ravioli (vegan)']))
# expected: 21.5
```

### Check — Part 1

```python
def check_part1():
    assert brunch is not None, "brunch not created yet"
    assert brunch.name == 'brunch'
    assert (brunch.start_time, brunch.end_time) == (11, 16), "brunch hours should be 11 and 16"
    assert (early_bird.start_time, early_bird.end_time) == (15, 18)
    assert (dinner.start_time, dinner.end_time) == (17, 23)
    assert (kids.start_time, kids.end_time) == (11, 21)
    assert repr(brunch) == 'brunch menu available from 11 am to 4 pm', f"got: {brunch!r}"
    assert brunch.calculate_bill(['pancakes', 'home fries', 'coffee']) == 13.5
    assert early_bird.calculate_bill(
        ['salumeria plate', 'mushroom ravioli (vegan)']) == 21.5
    print('Part 1 OK')

check_part1()
```

## Part 2 — Creating the Franchises

**Steps 13–15.** Create a `Franchise` class whose constructor takes `self, address, menus` and stores
both. `menus` is a list of `Menu` objects.

**Step 17.** Add a `__repr__` that returns the franchise address.

**Step 18.** Add `available_menus(self, time)` that returns a list of the menus served at that hour —
a menu counts if `start_time <= time <= end_time`.

```python
class Franchise:
  def __init__(self, address, menus):
    self.address = address
    self.menus = menus
  def __repr__(self):
    return self.address
  def available_menus(self, time):
    available = []
    for menu in self.menus:
      if menu.start_time <= time <= menu.end_time:
        available.append(menu)
    return available
```

**Step 16.** Create the two locations, both serving all four menus:

- `flagship_store` — `1232 West End Road`
- `new_installment` — `12 East Mulberry Street`

```python
menus = [brunch, early_bird, dinner, kids]
flagship_store = Franchise('1232 West End Road', menus)
new_installment = Franchise('12 East Mulberry Street', menus)
```

**Steps 19–20.** What's being served at noon? At 5 pm?

```python
print(flagship_store)
# expected: 1232 West End Road

print(flagship_store.available_menus(12))
# expected: [brunch menu ..., kids menu ...]

print(flagship_store.available_menus(17))
# expected: [early-bird menu ..., dinner menu ..., kids menu ...]
```

### Check — Part 2

```python
def check_part2():
    assert flagship_store is not None, "flagship_store not created yet"
    assert repr(flagship_store) == '1232 West End Road', f"got: {flagship_store!r}"
    assert repr(new_installment) == '12 East Mulberry Street'
    assert flagship_store.menus == menus
    noon = flagship_store.available_menus(12)
    assert [m.name for m in noon] == ['brunch', 'kids'], [m.name for m in noon]
    five = flagship_store.available_menus(17)
    assert [m.name for m in five] == ['early-bird', 'dinner', 'kids'], [m.name for m in five]
    print('Part 2 OK')

check_part2()
```

## Part 3 — Creating Businesses

**Steps 21–23.** Create a `Business` class whose constructor takes `self, name, franchises` and stores
both. Then create the business `"Basta Fazoolin' with My Heart"` with both franchises.

**Steps 24–25.** A second business — *Take a' Arepa* — reuses the same three classes:

- menu `Take a' Arepa`, served 10 am – 8 pm, items below
- franchise at `189 Fitzgerald Avenue` serving just that menu
- business `Take a' Arepa` with that one franchise

```python
class Business:
  def __init__(self, name, franchises):
    self.name = name
    self.franchises = franchises


basta = Business("Basta Fazoolin' with My Heart", [flagship_store, new_installment])
```

```python
arepas_items = {
  'arepa pabellon': 7.00, 'pernil arepa': 8.50, 'guayanes arepa': 8.00, 'jamon arepa': 7.50
}
arepas_menu = Menu("Take a' Arepa", arepas_items, 10, 20)
arepas_place = Franchise('189 Fitzgerald Avenue', [arepas_menu])
arepa = Business("Take a' Arepa", [arepas_place])
```

```python
print(basta.name, basta.franchises)
print(arepas_menu)
# expected: Take a' Arepa menu available from 10 am to 8 pm
print(arepas_place)
# expected: 189 Fitzgerald Avenue
```

### Check — Part 3

```python
def check_part3():
    assert basta is not None, "basta not created yet"
    assert basta.name == "Basta Fazoolin' with My Heart"
    assert basta.franchises == [flagship_store, new_installment]
    assert repr(arepas_menu) == "Take a' Arepa menu available from 10 am to 8 pm", f"got: {arepas_menu!r}"
    assert repr(arepas_place) == '189 Fitzgerald Avenue'
    assert arepas_place.menus == [arepas_menu]
    assert arepa.name == "Take a' Arepa" and arepa.franchises == [arepas_place]
    print('Part 3 OK')

check_part3()
```
