```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

#load and investigate the data here:
df = pd.read_csv("tennis_stats.csv")
#Look at Data of tennis players Step 2.) 
print(df)
```

```python
# perform exploratory analysis here:
#Step 3.) Look at relationships 2-3 features vs Outcomes 
# Aces Vs. Wins (1), #Aces Vs. Losses (2),  BreakPointsConverted Vs. Winnings (3) 

#Step 3) Acessing features and outcomes and Scatter Plot the relations
X_aces = df[["Aces"]]
Y_Wins = df[["Wins"]]
Y_Losses = df[["Losses"]]
x_Bp = df[["BreakPointsConverted"]]
Y_winnings = df[["Winnings"]]


#Plotting Section Which extension of step 3.) 
plt.figure()
plt.scatter(X_aces, Y_Wins)
plt.xlabel("Aces"); plt.ylabel("Wins")
plt.show()

plt.figure()
plt.scatter(X_aces, Y_Losses)
plt.xlabel("Aces"); plt.ylabel("Losses")
plt.show()

plt.figure()
plt.scatter(x_Bp, Y_winnings)
plt.xlabel("BreakPointsConverted"); plt.ylabel("Winnings")
plt.show()
```

```python
#Step 4.) Linear Regression on 1-feature and outcome: Aces Vs Wins
x_ac_train, x_ac_test, y_win_train, y_win_test = train_test_split(X_aces, Y_Wins, train_size = 0.8, test_size = 0.2)
LR_ac_w = LinearRegression()
model1 = LR_ac_w.fit(x_ac_train, y_win_train)
model1.score(x_ac_test, y_win_test)
Y_winP = model1.predict(x_ac_test)
plt.figure()
plt.scatter(y_win_test, Y_winP, alpha=0.4)
plt.xlabel("Win Test"); plt.ylabel("Win Predict")
```

```python
#Step 5.) Linear Regression on 1-feature and outcome: Aces Vs Losses
x_ac1_train, x_ac1_test, y_Los_train, y_Los_test = train_test_split(X_aces, Y_Losses, train_size = 0.8, test_size = 0.2)
LR_ac1_w = LinearRegression() 
model2 = LR_ac_w.fit(x_ac1_train, y_Los_train) 
model2.score(x_ac1_test, y_Los_test)
YLos_Pred = model2.predict(x_ac1_test)
plt.figure()
plt.scatter(y_Los_test, YLos_Pred, alpha=0.4)
plt.xlabel("Loss Test"); plt.ylabel("Loss Predict")
```

```python
#Step 6.) 2-features Vs. Outcome: Winnings  (Use: 'BreakPointsOpportunities', 'FirstServeReturnPointsWon' Vs. Winnings) 
Inp_Feat_S_6 = df[["BreakPointsOpportunities", "FirstServeReturnPointsWon"]]
Out_Feat_S_6 = df[["Winnings"]]
Inp_Train, Inp_Test, Out_Train, Out_Test = train_test_split(Inp_Feat_S_6, Out_Feat_S_6, train_size = 0.8, test_size = 0.2)
LR = LinearRegression() 
modelS6 = LR.fit(Inp_Train, Out_Train) 
modelS6.score(Inp_Test, Out_Test)
Y_Pred_Out = modelS6.predict(Inp_Test)
plt.figure()
plt.scatter(Out_Test, Y_Pred_Out, alpha=0.4)
plt.xlabel("Winnings Test"); plt.ylabel("Winnings Predict")
```

```python
#Step 7.) 2-features Vs. Outcome: Winnings  (Use: multi-Feature Vs. Winnings) 
Inp_Feat_S_7 = df[['FirstServe','FirstServePointsWon','FirstServeReturnPointsWon',
'SecondServePointsWon','SecondServeReturnPointsWon','Aces',
'BreakPointsConverted','BreakPointsFaced','BreakPointsOpportunities',
'BreakPointsSaved','DoubleFaults','ReturnGamesPlayed','ReturnGamesWon',
'ReturnPointsWon','ServiceGamesPlayed','ServiceGamesWon','TotalPointsWon',
'TotalServicePointsWon']]
Out_Feat_S_7 = df[["Winnings"]]
Inp_Train, Inp_Test, Out_Train, Out_Test = train_test_split(Inp_Feat_S_7, Out_Feat_S_7, train_size = 0.8, test_size = 0.2)
LR = LinearRegression() 
modelS7 = LR.fit(Inp_Train, Out_Train) 
modelS7.score(Inp_Test, Out_Test)
Y_Pred_Out = modelS7.predict(Inp_Test)
plt.figure()
plt.scatter(Out_Test, Y_Pred_Out, alpha=0.4)
plt.xlabel("Winnings Test"); plt.ylabel("Winnings Predict")
```
