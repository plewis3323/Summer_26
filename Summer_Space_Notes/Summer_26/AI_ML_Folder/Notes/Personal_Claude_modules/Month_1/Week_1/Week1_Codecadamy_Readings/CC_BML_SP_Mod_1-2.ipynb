 ## Machine Learning — Main Categories
  - ML branches into two main categories:
    - **Supervised Learning** — data is *labeled*
    - **Unsupervised Learning** — data is *unlabeled*

  ## Supervised Learning
  - The data is **labeled**: each input comes with its correct output.
  - The program learns to **predict the output from the input data**.
  - *Example — credit card fraud detection:*
    - **Input:** a set of recorded transactions
    - **Output:** for each transaction, predict *fraudulent* or *not fraudulent*
  - Splits into two sub-types: **Regression** and **Classification**.

  ## Regression
  - Goal: predict a **continuous-valued** output ($y \in \mathbb{R}$).
  - The output can take *any* value on a continuous range.
  - *Examples:*
    - What is the housing price in Neo York?
    - What is the value of cryptocurrencies?

  ## Classification
  - Goal: predict a **discrete** number of values (a category/class).
  - The output belongs to a *finite set* of labels, e.g. $y \in \{0, 1\}$.
  - *Examples:*
    - Is this a picture of a human or an AI?
    - Is this email spam or not spam?

  ## Quick Comparison
  | Type | Output | Example |
  |------|--------|---------|
  | **Regression** | Continuous ($\mathbb{R}$) | House price |
  | **Classification** | Discrete (classes) | Spam / Not spam |

 ## Unsupervised Learning
  - A type of machine learning where the program learns the **inherent structure** of the data.
  - Works from **unlabeled examples** — there is no "correct output" given.
  - Goal: discover hidden patterns/structure on its own (no target variable $y$).
  - Contrast with **supervised learning**, which learns from *labeled* input→output pairs.

  ## Clustering
  - A common unsupervised approach.
  - Finds patterns/structure in unlabeled data by **grouping points into clusters**.
  - Points in the same cluster are more *similar* to each other than to points in other clusters.
  - *Examples:*
    - Social networks clustering topics in a news feed
    - Consumer sites clustering users for recommendations
    - Search engines grouping similar objects into one cluster

  ## Supervised vs. Unsupervised — at a glance
  | | Supervised | Unsupervised |
  |---|---|---|
  | **Data** | Labeled | Unlabeled |
  | **Goal** | Predict output from input | Find inherent structure |
  | **Main tasks** | Regression, Classification | Clustering |
