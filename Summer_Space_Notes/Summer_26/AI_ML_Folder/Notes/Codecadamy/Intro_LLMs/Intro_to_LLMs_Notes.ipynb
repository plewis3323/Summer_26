## Lesson 1: Chatbots, Language Models and the Birth of AI

### ELIZA (1966) — the first digital chatbot
- Built at MIT by **Joseph Weizenbaum**
- **Rule-based**: followed a pre-determined "script" to assume a persona (no learning, no statistics)
- Famous persona: **DOCTOR**, a mock psychotherapist that mirrored user statements back ("I'm feeling X" → "What is making you feel X?")
- **ELIZA effect**: the psychological tendency to unconsciously anthropomorphize (attribute humanness to) computer behavior — named because people were so easily convinced ELIZA was human

### The Turing Test (1950)
- From **Alan Turing's** paper *"Computing Machinery and Intelligence"*
- Turing argued "Can machines think?" is meaningless; the better question: **can a machine imitate human behavior well enough to fool a human?**
- His **"imitation game"**: a human evaluator chats with a machine and a human via text only; the machine "passes" if the evaluator can't tell which is which
- Both ELIZA and ChatGPT can be said to pass it — a hint that the Turing test measures *convincingness*, not intelligence

### Dartmouth Workshop (1956) — the birth of AI as a field
- 6 years after Turing's paper
- Didn't define AI; instead stated the core **conjecture**: every feature of learning/intelligence can in principle be described precisely enough for a machine to simulate it
- Explicitly called out **language** as a key target: making machines "use language, form abstractions and concepts, solve problems reserved for humans, and improve themselves"

### The key shift (setup for the rest of the module)
- Old: **rule-based** chatbots (ELIZA) — hand-written rules
- New: **language models using neural networks** that capture **statistical regularities in language** — this is what makes modern LLMs like ChatGPT possible

## Lesson 2: Detecting Patterns in Text

### Historical roots of language modeling
- **1913 — A.A. Markov**: analyzed letter sequences in a Russian novel and showed language patterns can be captured mathematically (origin of Markov-style sequence analysis)
- **Claude Shannon** (decades later) framed the key question: **can we predict the next letter given a series of letters?**

### Next-letter / next-word prediction (the XKCD example)
- Given "eru-", only a few English words fit: *erudite*, *erupt*, *eructate*
- **Context narrows the choice**: the word "volcano" points to *erupt*; the present-continuous tense points to *erupting*
- Humans do this using internalized language patterns — NLP asks whether an algorithm can do it too

### Natural Language Processing (NLP)
- The field dedicated to automating language understanding/prediction with algorithms
- Core requirement: **turn text into math** — early steps in any NLP task involve standardizing text so computers can analyze it numerically

### Language units: n-grams and tokens

| Unit | Definition | Example from the "eru-" sentence |
|---|---|---|
| **Unigram** | single-letter unit | {o, h, _, m, y, ...} |
| **Bigram** | two-letter unit | {oh, h_, _m, my, ...} |
| **Trigram** | three-letter unit | {oh_, h_m, ...} |
| **n-gram** | n-letter (or n-word) unit | general case |
| **Token** | words, subwords (-ing, -ed), or word groups — the unit used by modern language models | "erupt" + "-ing" |

- Best unit length depends on the problem; there's no single right answer
- Words are a more "natural" unit than letters (split on spaces) — e.g., the **Google Books n-gram viewer** uses word units to show word frequency over time
- **Modern LLMs use tokens** (often subwords)

> **Big idea:** Formalizing language into units (n-grams/tokens) lets us build a model that predicts the next best unit — next-token prediction is the foundation of LLMs.

## Lesson 3: Autoregressive Language Models

### Common NLP tasks
- **Translation** (Google Translate)
- **Sequence completion / autocomplete**
- **Question-answering** (customer service chatbots)
- **Sentiment analysis** (content filters)

### What a "language model" is
- A single algorithmic system that can do *all* of these tasks and more — the unifying goal of NLP research
- The big hope: since language carries human thought, scaling up language models might produce something like generalized intelligence (an open question)

### The deceptively simple core question
> Given some text data, how do you build an algorithm to predict the next best thing to say?

A model that answers this well turns out to be capable of many other language tasks. This is *the* central insight of the whole module.

### Key vocab

| Term | Meaning |
|---|---|
| **Corpus** | The body of text data a model is built from (books, articles, websites) — usually gathered from many sources and digitally stored |
| **Autoregressive** | A statistical model that predicts future values from past values — for LLMs: using previous words/tokens to predict the next one |
| **Count-based language model** | The simplest language model: a giant lookup table of all word combinations in the corpus and their frequencies, used to look up how likely a given sentence is |

> **Big idea:** Modern LLMs (including GPT models) are autoregressive — they generate text one token at a time, each prediction conditioned on everything that came before. The count-based lookup table is the primitive ancestor of that idea.

## Lesson 4: Count-based Autoregressive Language Models

### How a count-based model scores a sentence
To find how likely "What do I say next?" is, ask in order:
1. How likely is "what"?
2. How likely is "do" *after* "what"?
3. How likely is "I" *after* "what do"? ... and so on.

### Conditional probability notation
- **P(do | what)** — read "P of *do* given *what*" — the probability that "do" appears given "what" already appeared
- Estimated straight from the lookup table by normalizing counts:

$$P(\text{do} \mid \text{what}) = \frac{40}{40+16+16+8} = \frac{40}{80} = 0.5$$

### Chain rule for sequences

$$P(\text{what do}) = P(\text{what}) \times P(\text{do} \mid \text{what})$$

And for the full sentence:

$$P(\text{sentence}) = P(\text{what}) \cdot P(\text{do}|\text{what}) \cdot P(\text{I}|\text{what do}) \cdot P(\text{say}|\text{what do I}) \cdot P(\text{next}|\text{what do I say})$$

> **Pattern to remember:** conditional probability = (count of prefix + word) ÷ (sum of counts of all words following that prefix); sentence probability = product of the conditionals.

### 📝 Worked Exercise: Does P(monkeys | what do) match the count table?

**Question:** Examine the counts of the next word to see if the probabilities match up. Does the probability of "monkeys" after "what do" match the count tables? Based on this, what would P(what do monkeys) be?

**Step 1 — Counts for the four possible next words after "what do":**

| Next word | Count |
|---|---|
| I | 8 |
| you | 6 |
| we | 5 |
| monkeys | 1 |
| **Total** | **20** |

**Step 2 — Conditional probability:**

$$P(\text{monkeys} \mid \text{what do}) = \frac{1}{8+6+5+1} = \frac{1}{20} = 0.05$$

✅ Matches the 0.05 in the tree diagram — the probabilities are just normalized counts.
(Check on the others: I = 8/20 = 0.4, you = 6/20 = 0.3, we = 5/20 = 0.25 — all match.)

**Step 3 — P(what do monkeys) via the chain rule:**

$$P(\text{what do monkeys}) = P(\text{what}) \times P(\text{do} \mid \text{what}) \times P(\text{monkeys} \mid \text{what do}) = P(\text{what}) \times 0.5 \times 0.05$$

Taking "What" as the given start of the sentence (P(what) = 1):

$$P(\text{what do monkeys}) = 0.5 \times 0.05 = \mathbf{0.025}$$

**Takeaway:** a count-based LM assigns sentence probability by multiplying conditionals (chain rule). This also previews the weakness: "What do I be" has count 0, so any sentence containing it gets probability 0 — count tables can't generalize to unseen word combinations.

```python
# Runnable version of the count-based exercise
counts_after_what = {"do": 40, "am": 16, "should": 16, "have": 8}
counts_after_what_do = {"I": 8, "you": 6, "we": 5, "monkeys": 1}

def conditional_probs(counts):
    total = sum(counts.values())
    return {word: count / total for word, count in counts.items()}

p_given_what = conditional_probs(counts_after_what)
p_given_what_do = conditional_probs(counts_after_what_do)

print("P(next | 'what')    :", p_given_what)
print("P(next | 'what do') :", p_given_what_do)

p_sentence = p_given_what["do"] * p_given_what_do["monkeys"]
print(f"\nP('what do monkeys') = {p_given_what['do']} * {p_given_what_do['monkeys']} = {p_sentence}")
```

## Lesson 5: Generalization — From Count-based to Neural Language Models

### The fatal flaw of count-based models
- "A lion is chasing a llama" — plausible in principle, but if that exact word sequence never appears in the training corpus, a count-based model assigns it **zero probability** (= "impossible")
- Root cause: count-based models **cannot generalize**

### Key vocab

| Term | Meaning |
|---|---|
| **Generalization** | A model's ability to adapt to new/unseen data — for language models, producing text that doesn't exist in its training data |
| **Semantic representation** | Linking words to their *meaning and context* rather than just counting occurrences (lion → predator, llama → prey/domesticated → "predators chase prey" → non-zero probability) |
| **Word embedding** | The mathematical object that enables generalization: each word becomes a series of numbers (a **word vector**) in an abstract space where semantically related words are linked |
| **Neural language model** | A language model whose embeddings are built using neural networks — the most popular and efficient way to do it |

### Why embeddings are more powerful — homonyms
- "lead" next to "President" vs. "lead" next to "pipe" — same spelling, different meanings
- An embedding-based model can distinguish them from context; a count-based model can't

### Two distinctions worth keeping straight (quiz material)
- **Generative** = can *generate* data (text/image/audio/video)
- **Generalizable** = can adapt to / produce *unseen* data
- **Zero probability** = "it is impossible that…" / **non-zero probability** = "it is possible that…"

> **Big idea:** counting tells you *which* words appeared; embeddings capture *what words mean*. Moving from counts to meaning is the leap that lets language models say new things — and neural networks are the machinery for that leap.

## Lesson 6: Compression — Solving the Curse of Dimensionality

### The curse of dimensionality
- As the corpus grows, the count table explodes
- Example: 100-word vocabulary, all possible 5-word sentences → **100⁵ = 10 billion counts** to store
- Giant count tables demand enormous memory and compute — it doesn't scale

### The fix: compression by approximation
- Neural language models **learn an approximation of the count table** instead of storing the whole thing
- The compressed form lives in the model's **parameters** (a.k.a. **weights**)
- Analogy: shrinking a high-res photo to low-res — easier to store; how much important detail survives depends on how clever the compression is (Ted Chiang's New Yorker piece describes ChatGPT as "a blurry JPEG of the web")

### Information loss — yes, it's real
- Compression means the model can sometimes assign **zero probability to text that actually exists** in the corpus
- But that same lossiness is what *enables* generalization: the model keeps semantic connections (lion → predator, llama → prey, predators chase prey) instead of exact strings, so it can assign non-zero probability to never-seen sentences

### The summary worth memorizing
Neural language models:
- can assign **zero** probabilities to **existing** text → that's **compression** (information loss)
- can assign **non-zero** probabilities to **unseen** text → that's **generalization**

> **Compression and generalization are two sides of the same coin.** A neural LM trades perfect recall of the training data for the ability to say new things. The blur is the feature.

## Lesson 7: Neural Networks and Language Models

### Symbolic vs. subsymbolic AI

| | Symbolic AI | Subsymbolic AI |
|---|---|---|
| Approach | Explicit rules grounded in mathematical logic | Learn underlying patterns; no explicit rules |
| Models | Conscious thought processes | Unconscious processes (perception, cognition) |
| Example | **ELIZA** (1966) — scripted responses | **Neural networks** |

- 1950s neuroscience showed unconscious processes drive much of perception/cognition — and these can't be captured by rule-making, which inspired the subsymbolic approach
- **1958 — Frank Rosenblatt's perceptron**: a simple program simulating a single neuron; the idea was that *layering* networks of perceptrons would produce emergent intelligent processing

### Machine learning and deep learning
- Neural nets went in and out of favor for ~50 years until the **early-2000s "age of data"** (internet → more data + compute)
- **Machine learning**: algorithms that learn patterns from data to do prediction, inference, generation
- **Deep learning**: the subset of ML using *deep* (many-layered) neural networks

### Neural network architectures (arrangement determines what the network is good at)
- **CNNs** (Convolutional Neural Networks) → image data
- **RNNs** (Recurrent Neural Networks) → sequential data like language; dominant in NLP (translation, speech recognition) from the mid-2000s
- **Transformer (2017)** → exceptional at language tasks; the architecture behind modern LLMs
- **GPT = Generative Pre-trained Transformer**: LLMs that use transformers trained on vast corpora to generate coherent human-sounding text

### ⭐ Extra context on transformers (beyond the lesson)

- Introduced in the 2017 Google paper **"Attention Is All You Need"** (Vaswani et al.)
- The key innovation is **self-attention**: every token can directly "look at" every other token in the input and weigh how relevant each one is. This is how the model resolves things like which meaning of "lead" applies — context words get high attention weights
- **Why they beat RNNs:**
  - RNNs process tokens one at a time, so they're slow to train and tend to "forget" distant context. Transformers process the whole sequence **in parallel**, making them dramatically faster to train on GPUs — which is what made *scaling* to billions of parameters practical
  - Attention handles **long-range dependencies** (a word at position 5 relating to one at position 500) far better than recurrence
- GPT-style models use the **decoder-only** variant of the transformer and are **autoregressive** — exactly the next-token-prediction setup from Lesson 3, just with a vastly better function approximator than a count table
- "Pre-trained" in GPT means the model first learns general language patterns from a huge corpus before any task-specific tuning

> **Big idea:** the module's arc completes here — rules (ELIZA) → counts → neural compression/embeddings → and the transformer is the specific architecture that made the neural approach scale into modern LLMs.

## Lesson 8: LLMs — Caveats and Possibilities

### Recap: what applies to all language models (small or large)
1. Trained on text corpora; outputs are shaped by what is/isn't in the training data
2. Most tasks reduce to **predicting the next best word/token**
3. Predictions come from **probability distributions** learned by neural networks
4. Text → **embeddings** (mathematical representations preserving meaning and context)
5. Neural LMs solve generalization + curse of dimensionality via **compression** into a semantic representation
6. Compression enables generalization *and* causes information loss

### What changes at large scale → Emergence
- **Emergence** = a sudden, sharp increase in model performance as scale grows
- Borrowed from physics (Phillip Anderson): *"quantitative changes in a system result in qualitative changes in behavior"*
- In LLMs: abilities nobody trained for show up — joke explanation, counterfactuals, imitating author styles, writing poetry
- This fuels speculation about **AGI (Artificial General Intelligence)** — a generalized emergent intelligence

### Hallucinations
- **Hallucination** = the model producing incorrect/erroneous text
- It's the *flip side of generalization*: generated unseen text is, by definition, unverified — it didn't exist before, so no human has checked it

### Why factual grounding is structurally hard for LLMs
- The model doesn't know what's factual — it relies on the veracity of its training text
- Missing information → generalization fills the gap with the *most plausible thing to say*, which may be reasonable, absurd, or simply untrue
- **Compression** makes it worse (information loss), but even a lossless model would still be:
  - **static** — can't incorporate new facts or changing information
  - **unable to know what it doesn't know**
- Modern models (GPT-4, PaLM) use **human feedback** to reduce hallucinations, but they **cannot be completely eliminated** — it's inherent to how the models are built

> **Big idea:** emergence and hallucination are both consequences of the same machinery. Generalization at scale gives you poetry *and* confident falsehoods — you can't get one without the risk of the other.

## Lesson 9: LLM Parameters — Temperature

### What temperature is
- A parameter (available in chat UIs and APIs) that controls the **sharpness of the probability distribution** the model generates over next tokens
- It reshapes the distribution: boosting high-probability outcomes and suppressing low-probability ones (narrower), or the reverse (wider)

### Low temperature → narrower distribution
- High-probability words get even more likely; low-probability words get squeezed out
- Example: for input "The cat", the already-likely "chases" becomes near-certain
- Result: **more deterministic** output — the model tends to produce the same outcome every time

### High temperature → wider distribution
- Lower-probability outcomes become viable — "eats" might beat "chases"; push it further and "eats mango" comes into play
- Result: **more novel/creative/random** output — this is what API docs mean by "tunes for more creative outcomes"

### Quick reference

| Temperature | Distribution | Output behavior | Use when |
|---|---|---|---|
| Low | Narrow/sharp | Deterministic, repeatable | Factual Q&A, code, consistency |
| High | Wide/flat | Varied, creative, surprising | Brainstorming, fiction, variety |

### ⚠️ The caveat (quiz material)
> **Low temperature is no guarantee of correctness.** High probability ≠ factually grounded — temperature changes *which* plausible token gets picked, not whether the model knows the truth. Always verify LLM output against facts.

- Other parameters exist too (frequency penalties, sampling settings, etc.) — covered in Codecademy's LLM API course

**Connection to earlier lessons:** the model always produces a distribution over next tokens (Lessons 3–4); temperature just decides how adventurously you sample from it.

```python
# Demo: how temperature reshapes a next-token probability distribution
import math

# Toy distribution for the prompt "The cat ..."
base_probs = {"chases": 0.5, "sleeps": 0.25, "eats": 0.15, "eats mango": 0.10}

def apply_temperature(probs, temperature):
    # Rescale log-probabilities by temperature, then re-normalize (softmax)
    scaled = {w: math.exp(math.log(p) / temperature) for w, p in probs.items()}
    total = sum(scaled.values())
    return {w: v / total for w, v in scaled.items()}

for t in [0.2, 1.0, 2.0]:
    adjusted = apply_temperature(base_probs, t)
    label = "low (deterministic)" if t < 1 else ("baseline" if t == 1 else "high (creative)")
    print(f"Temperature {t} — {label}")
    for word, p in adjusted.items():
        print(f"  {word:<11} {p:.3f} {'█' * int(p * 40)}")
    print()
```

*Why:* Raising temperature flattens the distribution, so it's specifically the **lower**-probability words that gain ground — that's what makes output more varied. "Less deterministic" by definition means the same prompt yields **different** responses across attempts.

## Key Takeaways

1. **Everything is next-token prediction.** From count tables to GPT-4, the core task is: given the previous tokens, predict the next best one (autoregressive modeling).
2. **Counts → meaning.** Count-based models can only repeat what they've seen (zero probability for unseen text). Word embeddings give words *semantic* representations, enabling generalization.
3. **Compression and generalization are two sides of the same coin.** Compressing a corpus into weights loses information but creates the semantic links that let the model say new things.
4. **Emergence:** at large scale, quantitative growth produces qualitative new abilities nobody trained for.
5. **Hallucination is structural.** Generated unseen text is by definition unverified; LLMs don't know what they don't know. Human feedback reduces but cannot eliminate hallucinations.
6. **Temperature tunes the sampling, not the truth.** Low = deterministic, high = creative — neither guarantees factual correctness. Always verify.
