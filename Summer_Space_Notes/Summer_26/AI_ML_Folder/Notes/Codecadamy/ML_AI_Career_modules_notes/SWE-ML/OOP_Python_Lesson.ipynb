## Introduction to Classes

### Types

Python stores data as different **types** (`str`, `int`, `float`, `list`, `dict`, …). A type decides what operations are legal: you can `.get()` on a dict, not on an int; you can `+` two ints, not two dicts. Those operations are defined **at the type level**.

Check a variable's type with `type()`.

````python
a_string = "Cool String"
an_int = 12

print(type(a_string))
# -> <class 'str'>

print(type(an_int))
# -> <class 'int'>
````

### Methods

**Methods** are functions defined on a class. The first parameter is always the object calling the method — conventionally named `self`. Methods always have at least this one parameter.

You don't pass `self` yourself. `pipi_pitbull.time_explanation()` automatically sends `pipi_pitbull` in as `self`.

````python
class Dog:
  dog_time_dilation = 7

  def time_explanation(self):
    print("Dogs experience {} years for every 1 human year.".format(self.dog_time_dilation))

pipi_pitbull = Dog()
pipi_pitbull.time_explanation()
# -> Dogs experience 7 years for every 1 human year.
````

### Methods With Arguments

Methods can take extra parameters after `self`. You still only pass those extras on the call — `self` is implicit.

`how_many_kms` is defined with `(self, miles)` but called as `converter.how_many_kms(5)`.

````python
class DistanceConverter:
  kms_in_a_mile = 1.609
  def how_many_kms(self, miles):
    return miles * self.kms_in_a_mile

converter = DistanceConverter()
kms_in_5_miles = converter.how_many_kms(5)
print(kms_in_5_miles)
# -> 8.045
````

### Constructors

Some methods have special behavior — **dunder methods** (double underscores on both sides).

`__init__` runs **every time** you instantiate. In Python, "constructor" usually means `__init__`. Arguments you pass to `ClassName(...)` land in `__init__` (after `self`).

````python
class Shouter:
  def __init__(self):
    print("HELLO?!")

shout1 = Shouter()
# -> HELLO?!

shout2 = Shouter()
# -> HELLO?!
````

````python
class Shouter:
  def __init__(self, phrase):
    # make sure phrase is a string
    if type(phrase) == str:

      # then shout it out
      print(phrase.upper())

shout1 = Shouter("shout")
# -> SHOUT

shout2 = Shouter("shout")
# -> SHOUT

shout3 = Shouter("let it all out")
# -> LET IT ALL OUT
````

### Instance Variables

Class variables are shared. **Instance variables** are data attached to one specific object — two instances of the same class can hold different values.

You can set them after the fact with the same `object.variable` notation: `fake_dict1.fake_key = "This works!"`.

````python
class FakeDict:
  pass

fake_dict1 = FakeDict()
fake_dict2 = FakeDict()

fake_dict1.fake_key = "This works!"
fake_dict2.fake_key = "This too!"

# Let's join the two strings together!
working_string = "{} {}".format(fake_dict1.fake_key, fake_dict2.fake_key)
print(working_string)
# -> This works! This too!
````

### Attribute Functions

Class variables and instance variables are both **attributes**. Accessing one that doesn't exist raises `AttributeError`.

- `hasattr(obj, "name")` → `True`/`False`
- `getattr(obj, "name", default)` → the value, or `default` if missing (third arg optional; without it, missing → `AttributeError`)

````python
class NoCustomAttributes:
  pass

attributeless = NoCustomAttributes()

try:
  attributeless.fake_attribute
except AttributeError:
  print("This text gets printed!")
# -> This text gets printed!

print(hasattr(attributeless, "fake_attribute"))
# -> False

print(getattr(attributeless, "other_fake_attribute", 800))
# -> 800
````

**Exercise** — `hasattr(element, "count")` on a dict, str, int, and list. `str` and `list` have `.count`; `dict` and `int` don't.

````python
can_we_count_it = [{'s': False}, "sassafrass", 18, ["a", "c", "s", "d", "s"]]

for element in can_we_count_it:
  if hasattr(element, "count"):
    print(str(type(element)) + " has the count attribute!")
  else:
    print(str(type(element)) + " does not have the count attribute :(")
# -> <class 'dict'> does not have the count attribute :(
# -> <class 'str'> has the count attribute!
# -> <class 'int'> does not have the count attribute :(
# -> <class 'list'> has the count attribute!
````

### Self

`self` is **this object**, not the class. The usual pattern: constructor arguments become instance variables via `self.url = url`. Then every instance has the same *shape* of data, with its own values.

Methods use `self` to reach both class variables (`self.secure_prefix`) and instance variables (`self.url`). That's the point of OOP: structure the data, then write methods that work with it.

````python
class SearchEngineEntry:
  def __init__(self, url):
    self.url = url

codecademy = SearchEngineEntry("www.codecademy.com")
wikipedia = SearchEngineEntry("www.wikipedia.org")

print(codecademy.url)
# -> www.codecademy.com

print(wikipedia.url)
# -> www.wikipedia.org
````

````python
class SearchEngineEntry:
  secure_prefix = "https://"
  def __init__(self, url):
    self.url = url

  def secure(self):
    return "{prefix}{site}".format(prefix=self.secure_prefix, site=self.url)

codecademy = SearchEngineEntry("www.codecademy.com")
wikipedia = SearchEngineEntry("www.wikipedia.org")

print(codecademy.secure())
# -> https://www.codecademy.com

print(wikipedia.secure())
# -> https://www.wikipedia.org
````

### String Representation

Default `print(obj)` is almost useless: `<__main__.Employee object at 0x...>`.

`__repr__(self)` is the dunder that controls that string. One parameter (`self`), must **return a string**. After you define it, `print(argus)` can show something you actually care about (e.g. the name).

````python
class Employee():
  def __init__(self, name):
    self.name = name

argus = Employee("Argus Filch")
print(argus)
# -> <__main__.Employee object at 0x...>  (default, not useful)
````

````python
class Employee():
  def __init__(self, name):
    self.name = name

  def __repr__(self):
    return self.name

argus = Employee("Argus Filch")
print(argus)
# -> Argus Filch
````

**Exercise** — `Circle.__repr__` returns `Circle with radius {radius}`. Print the three circles.

````python
class Circle:
  pi = 3.14

  def __init__(self, diameter):
    self.radius = diameter / 2

  def area(self):
    return self.pi * self.radius ** 2

  def circumference(self):
    return self.pi * 2 * self.radius

  def __repr__(self):
    return "Circle with radius {radius}".format(radius=self.radius)

medium_pizza = Circle(12)
teaching_table = Circle(36)
round_room = Circle(11460)

print(medium_pizza)
print(teaching_table)
print(round_room)
# -> Circle with radius 6.0
# -> Circle with radius 18.0
# -> Circle with radius 5730.0
````

**Q: What is `__init__()`?**

**A:** It's the class **constructor** — a dunder method that Python calls automatically every time you instantiate (`MyClass()`). Use it to set up the new object (store arguments, set starting state).

- Written with **two** underscores on each side: `__init__`, not `_init_`.
- First parameter is still `self` (the new object). Extra args you pass to `MyClass(...)` land here.
- You almost never call `__init__` yourself; `Shouter("shout")` does it.

**Q: What is a dunder method?**

**A:** A method named with **d**ouble **under**scores on both sides, like `__init__` or `__str__`. Also called a magic method.

Python calls them for you in special situations — you rarely call them by name:
- `__init__` → when you write `MyClass()`
- `__str__` → when you `print(obj)`
- `__add__` → when you write `a + b`
- `__len__` → when you write `len(obj)`

`dir(obj)` shows a bunch of these. The underscores are a "hands off, Python owns this" signal; don't invent your own `__foo__` names.

**Q: What does the `.format()` method do?**

**A:** It's a string method that fills in placeholders. The string has holes (`{}` or `{name}`); `.format(...)` plugs values into those holes and returns a new string.

```
"Dogs experience {} years for every 1 human year.".format(7)
# -> Dogs experience 7 years for every 1 human year.

"{prefix}{site}".format(prefix="https://", site="www.codecademy.com")
# -> https://www.codecademy.com

"Creating circle with diameter {d}".format(d=36)
# -> Creating circle with diameter 36
```

- `{}` fills in order of the arguments.
- `{name}` fills by keyword (`d=36`).
- Same idea as an f-string: `f"Creating circle with diameter {diameter}"` — the Circle constructor exercise used that form too.

**Q: What actually is the `__init__` constructor?**

**A:** It's the setup function for a new object. When you write `Circle(12)`, Python does two things:

1. Makes a blank `Circle` object.
2. Calls `__init__` on it, passing that object as `self` and `12` as `diameter`.

So this:

```
def __init__(self, diameter):
    self.radius = diameter / 2
```

means: "on *this* new circle, store a radius that's half of whatever diameter I was given."

After `medium_pizza = Circle(12)`, `medium_pizza.radius` is `6.0` because `__init__` put it there. Without `__init__`, you'd have an empty object and would have to set `.radius` by hand (like the `Store.store_name` exercise).

It's `__init__` (two underscores each side), not `_init_`. You don't call it — `Circle(12)` does.
